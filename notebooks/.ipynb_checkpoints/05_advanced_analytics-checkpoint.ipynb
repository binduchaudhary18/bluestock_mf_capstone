```python
nav_history['date'] = pd.to_datetime(nav_history['date'])

transactions['transaction_date'] = pd.to_datetime(
    transactions['transaction_date']
)

holdings['portfolio_date'] = pd.to_datetime(
    holdings['portfolio_date']
)
```

```python
nav_history = nav_history.sort_values(
    ['amfi_code','date']
)
```

```python
nav_history['daily_return'] = (
    nav_history.groupby('amfi_code')['nav']
    .pct_change()
)
```

```python
nav_history.head()
```

```python
def var_cvar(x):

    var95 = x.quantile(0.05)

    cvar95 = x[x <= var95].mean()

    return pd.Series({
        'VaR_95': var95,
        'CVaR_95': cvar95
    })
```

```python
var_report = (
    nav_history
    .groupby('amfi_code')['daily_return']
    .apply(var_cvar)
    .reset_index()
)
```

```python
var_report.head()
```

```python
var_report.to_csv(
    'var_cvar_report.csv',
    index=False
)
```

```python
risk_funds = var_report.sort_values(
    'CVaR_95'
)

risk_funds.head(10)
```

```python
print(var_report.head())
print(var_report.columns)
```

```python
var_report = (
    nav_history
    .groupby('amfi_code')['daily_return']
    .apply(var_cvar)
    .unstack()
    .reset_index()
)
```

```python
print(var_report.columns)
```

```python
risk_funds = var_report.sort_values('CVaR_95')
risk_funds.head(10)
```

```python
returns_df = nav_history.pivot(
    index='date',
    columns='amfi_code',
    values='daily_return'
)
```

```python
rolling_sharpe = (
    returns_df.rolling(90).mean()
    /
    returns_df.rolling(90).std()
) * np.sqrt(252)
```

```python
top5 = returns_df.columns[:5]
```

```python
plt.figure(figsize=(12,6))

for fund in top5:
    plt.plot(
        rolling_sharpe.index,
        rolling_sharpe[fund],
        label=str(fund)
    )

plt.legend()
plt.title("90 Day Rolling Sharpe Ratio")
plt.xlabel("Date")
plt.ylabel("Sharpe Ratio")

plt.savefig(
    "rolling_sharpe_chart.png"
)

plt.show()
```

```python
first_txn = (
    transactions
    .groupby('investor_id')
    ['transaction_date']
    .min()
)
```

```python
transactions['cohort_year'] = (
    transactions['investor_id']
    .map(first_txn.dt.year)
)
```

```python
avg_investment = (
    transactions
    .groupby('cohort_year')
    ['amount_inr']
    .mean()
)
```

```python
top_fund = (
    transactions
    .groupby(
        ['cohort_year','amfi_code']
    )
    .size()
    .reset_index(name='count')
)
```

```python
top_pref = (
    top_fund
    .sort_values(
        ['cohort_year','count'],
        ascending=[True,False]
    )
    .groupby('cohort_year')
    .head(1)
)
```

```python
sip = transactions[
    transactions['transaction_type']
    .str.contains(
        'sip',
        case=False,
        na=False
    )
].copy()
```

```python
sip = sip.sort_values(
    ['investor_id',
     'transaction_date']
)
```

```python
sip['gap_days'] = (
    sip.groupby('investor_id')
    ['transaction_date']
    .diff()
    .dt.days
)
```

```python
sip_count = (
    sip.groupby('investor_id')
    .size()
)
```

```python
eligible = sip_count[
    sip_count >= 6
].index
```

```python
sip = sip[
    sip['investor_id']
    .isin(eligible)
]
```

```python
avg_gap = (
    sip.groupby('investor_id')
    ['gap_days']
    .mean()
)
```

```python
at_risk = avg_gap[
    avg_gap > 35
]
```

```python
len(at_risk)
```

```python
sip_continuity_rate = (
    1 -
    len(at_risk)/len(avg_gap)
) * 100

sip_continuity_rate
```

```python
sharpe = (
    nav_history.groupby('amfi_code')['daily_return'].mean()
    /
    nav_history.groupby('amfi_code')['daily_return'].std()
) * np.sqrt(252)

sharpe = sharpe.reset_index()
sharpe.columns = ['amfi_code', 'sharpe_ratio']
```

```python
recommend_df = pd.merge(
    fund_master,
    sharpe,
    on='amfi_code'
)
```

```python
recommend_df.head()
```

```python
def recommend(risk):

    return (
        recommend_df[
            recommend_df['risk_category'] == risk
        ]
        .sort_values(
            'sharpe_ratio',
            ascending=False
        )[
            [
                'scheme_name',
                'risk_category',
                'sharpe_ratio'
            ]
        ]
        .head(3)
    )
```

```python
recommend("Very High")
```

```python
sector_weights = (
    holdings
    .groupby(
        ['amfi_code','sector']
    )['weight_pct']
    .sum()
    .reset_index()
)
```

```python
hhi = (
    sector_weights
    .assign(
        sq=lambda x:
        (x['weight_pct']/100)**2
    )
    .groupby('amfi_code')
    ['sq']
    .sum()
    .reset_index()
)
```

```python
hhi.columns = [
    'amfi_code',
    'HHI'
]
```

```python
hhi.sort_values(
    'HHI',
    ascending=False
)
```

# Key Insights

### 1. Risk Analysis Using VaR and CVaR

The downside risk analysis revealed that Fund 101207 exhibited the highest risk among all analyzed schemes, with a VaR of -2.60% and a CVaR of -3.25%. This indicates that during adverse market conditions, the fund experiences relatively larger losses compared to other schemes.

### 2. Investor Cohort Investment Patterns

The cohort analysis showed that investors who began investing in 2025 contributed a slightly higher average investment amount (₹109,159) compared to investors from the 2024 cohort (₹107,423). This suggests growing investor confidence and higher participation levels among newer investors.

### 3. Fund Preference Across Cohorts

Among investors who started investing in 2024, AMFI Code 148568 emerged as the most preferred scheme with 874 transactions. This indicates strong investor interest and popularity of the fund within the cohort.

### 4. SIP Continuity and Investor Retention

The SIP continuity analysis identified 1,332 investors as at-risk due to average SIP gaps exceeding 35 days. Such investors may require targeted engagement strategies to improve investment consistency and reduce the likelihood of SIP discontinuation.

### 5. Portfolio Concentration Analysis

The HHI analysis showed that Fund 119092 had the highest concentration score (0.297), indicating greater exposure to a limited number of sectors. Funds with higher HHI values may carry higher concentration risk compared to more diversified portfolios.


# Conclusion

This project performed advanced analytics on mutual fund and investor transaction data to evaluate fund risk, investor behavior, and portfolio characteristics. Historical VaR and CVaR metrics were used to assess downside risk across schemes, while rolling Sharpe ratio analysis helped evaluate risk-adjusted performance over time.

Investor cohort analysis provided insights into investment trends and fund preferences, revealing differences in behavior across investor groups. SIP continuity analysis highlighted a significant number of investors with irregular contribution patterns, which may impact long-term wealth creation. The recommendation engine demonstrated how funds can be matched with investor risk profiles using quantitative performance metrics.

Finally, the Herfindahl-Hirschman Index (HHI) was used to measure portfolio concentration and identify funds with greater sector exposure risk. Overall, the analysis illustrates how data-driven techniques can support mutual fund evaluation, investor segmentation, portfolio monitoring, and investment decision-making.
