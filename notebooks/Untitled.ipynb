```python
import pandas as pd
import os

data_path = "../data/raw"

datasets = {}

for file in sorted(os.listdir(data_path)):
    if file.endswith(".csv"):

        df = pd.read_csv(os.path.join(data_path, file))

        datasets[file] = df

        print("=" * 80)
        print("FILE:", file)

        print("\nShape:")
        print(df.shape)

        print("\nData Types:")
        print(df.dtypes)

        print("\nFirst 5 Rows:")
        print(df.head())

        print("\n")
```

```python
for file, df in datasets.items():

    print("=" * 80)
    print("FILE:", file)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\n")
    
```

```python
summary = []

for file, df in datasets.items():

    missing = df.isnull().sum().sum()
    duplicates = df.duplicated().sum()

    summary.append({
        "file": file,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "missing_values": missing,
        "duplicate_rows": duplicates
    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
summary_df.to_csv(
    "../reports/data_quality_summary.csv",
    index=False
)

print("Summary saved successfully!")
```

```python
fund_master = datasets["01_fund_master.csv"]
nav_history = datasets["02_nav_history.csv"]

master_codes = set(fund_master["amfi_code"])
nav_codes = set(nav_history["amfi_code"])

missing_codes = master_codes - nav_codes

print("Total AMFI codes in fund_master:", len(master_codes))
print("Total AMFI codes in nav_history:", len(nav_codes))
print("Missing AMFI codes:", len(missing_codes))

if len(missing_codes) > 0:
    print("\nMissing Codes:")
    print(missing_codes)
else:
    print("\n All AMFI codes from fund_master exist in nav_history")
```
